###  <span style="color:#D8E983" > 單一份年報抓取 : 以TSMC為例 </span>
#### <span style="color:#FFFFFF" >若存成.xlsx檔 --> 會出現問題 </span>
#### <span style="color:#B7BDF7" > 原因 1 : 可能出現禁止字元： 不能包含 [ ] : * ? / \ 這些符號。 </span>
#### <span style="color:#B7BDF7" > 原因 2 : 長度限制： 最多只能有 31 個字元 / \ 這些符號。 </span>

#### <span style="color:#F2A65A" > .csv檔 v.s .xlsx檔差異  </span>
#### <span style="color:#F2A65A" > 1. csv檔佔用空間較少，且為純文本格式  </span>


```python
import pandas as pd
all_data=[]
output_csv = "2330_年報.csv" 
try:
    with pdfplumber.open(path_2330) as pdf:
        total_pages = len(pages)
        print(f"這份文件共有 {total_pages} 頁")

        # 2. 迴圈遍歷每一頁 enumerate--> (0,data1) (1,data2)
        for i, page in enumerate(pages):
            # 抓取該頁文字
            text = page.extract_text()
            
            # 如果該頁有文字，就存下來；如果是空白頁則存 "無內容"
            content = text if text else "無內容"
            
            # 把「頁碼」跟「內容」存成一個小清單
            # i + 1 是因為程式是從 0 開始數，但我們習慣看第 1 頁
            all_data.append([i + 1, content])
            
            # 每處理 10 頁印一次進度，讓你知道程式還在跑
            if (i + 1) % 10 == 0:
                print(f"已處理 {i + 1} / {total_pages} 頁...")

    # 3. 將 List 轉換成 Pandas DataFrame
    df = pd.DataFrame(all_data, columns=["頁碼", "內容"])

    # 4. 存成 Excel
    # engine='openpyxl' 是為了確保相容性
    df.to_csv(output_csv, index=False, encoding='utf-8-sig',errors="ignore")
    
    print("-" * 30)
    print(f"成功！檔案已儲存為：{output_csv}")
    print("你可以打開 Excel 使用 Ctrl+F 搜尋關鍵字了。")

except FileNotFoundError:
    print("錯誤：找不到檔案，請確認路徑是否正確。")
except Exception as e:
    print(f"發生未預期的錯誤：{e}")
    pass
```

```python
import time
import os
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

# --- 1. 設定瀏覽器選項 (包含自動下載設定) ---
download_dir = os.path.join(os.getcwd(), "downloads") # 設定下載到當前目錄下的 downloads 資料夾
if not os.path.exists(download_dir):
    os.makedirs(download_dir)

chrome_options = Options()
# chrome_options.add_argument("--headless")  # 如果想看過程，這行先註解掉
chrome_options.add_argument("--start-maximized")
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")
# 加入這兩行忽略憑證錯誤
chrome_options.add_argument('--ignore-certificate-errors')
chrome_options.add_argument('--ignore-ssl-errors')

# 設定 Chrome 自動下載 PDF，不要跳出預覽視窗
prefs = {
    "download.default_directory": download_dir,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True,
    "plugins.always_open_pdf_externally": True  # 關鍵：直接下載 PDF，不要在瀏覽器開啟
}
chrome_options.add_experimental_option("prefs", prefs)





# --- 2. 啟動瀏覽器 ---
driver = webdriver.Chrome(options=chrome_options)
wait = WebDriverWait(driver, 10)

try:
    print("🚀 前往公開資訊觀測站 - 電子書專區...")
    url = "https://mops.twse.com.tw/mops/web/t57sb01_q5"
    driver.get(url)

    # --- 3. 輸入搜尋條件 ---
    print("⌨️  輸入代號 2330 與 年度 113...")
    
    # 輸入公司代號
    input_co_id = wait.until(EC.element_to_be_clickable((By.ID, "co_id")))
    input_co_id.clear()
    input_co_id.send_keys("2330")
    
    # 輸入年度 (113)
    input_year = driver.find_element(By.ID, "year")
    input_year.clear()
    input_year.send_keys("113")
    
    # 點擊搜尋按鈕
    # 注意：觀測站的按鈕有時候沒有 ID，用 input[type='button'] 比較保險
    search_btn = driver.find_element(By.XPATH, "//input[@type='button' and @value=' 查詢 ']")
    search_btn.click()

    # --- 4. 等待並抓取結果 ---
    print("⏳ 等待搜尋結果...")
    
    # 等待表格出現 (尋找包含 '電子書' 字樣的表格)
    wait.until(EC.presence_of_element_located((By.XPATH, "//table[contains(@class, 'hasBorder')]")))

    # 找到所有資料列 (Tr)
    # 邏輯：找到包含「年報」文字的那一列，然後去按那一列的按鈕
    target_row_xpath = "//tr[contains(., '年報') and not(contains(., '英文版'))]"
    
    # 這裡我們找 "中文版" 的 "年報" (避開英文版或議事手冊)
    # 如果找不到，會拋出 TimeoutException
    row = wait.until(EC.presence_of_element_located((By.XPATH, target_row_xpath)))
    
    print("✅ 找到年報資料列！")
    
    # 在這一列裡面，找到下載按鈕 (通常是一個 form submit input)
    # 觀測站的結構通常是 form > input[type='button']
    download_btn = row.find_element(By.XPATH, ".//input[@type='button' and @value='電子書']")
    
    # 點擊下載
    print("📥 正在點擊下載按鈕...")
    download_btn.click()
    
    # 等待下載完成 (簡單用 sleep 模擬，正規做法是檢查資料夾檔案)
    print(f"✅ 下載指令已送出！檔案將儲存於: {download_dir}")
    time.sleep(10) 

except Exception as e:
    print(f"❌ 發生錯誤: {e}")

finally:
    driver.quit()
    print("👋 瀏覽器已關閉")
```
