### 用requests模組 + bs4成功下載年報
### 第1個post會取得的結果 : https://doc.twse.com.tw/server-java/t57sb01?co_id=2330&year=114&colorchg=1&step=1&mtype=F

### requests模組官方文件 : https://requests.readthedocs.io/projects/cn/zh-cn/latest/user/quickstart.html

```python
%pip install lxml
import requests
from bs4 import BeautifulSoup
data={'id': '',
      'key': '',
      'step': '1',
      'co_id': '2330',
      'year': '114',
      'seamon': '',
      'mtype': 'F',
      'dtype': 'F04'}    

url='https://doc.twse.com.tw/server-java/t57sb01'  
# 第一個post請求
res=requests.post(url, data=data) 
soup=BeautifulSoup(res.text, "html.parser") 
link=soup.find('a')    
print("第1個link:",link)
filename=link.text
print("檔案名稱:",filename)

#第二個post請求
data={'step': '9',
      'kind': 'F',
      'co_id': '2330',
      'filename': filename} 

res=requests.post(url, data=data)   
soup=BeautifulSoup(res.text, 'html.parser')
link=soup.find('a') 
print("第2個link:",link)
href=link.get('href')
print("超連結:",href) 
pdf_url='https://doc.twse.com.tw' + href   

print("網址:",pdf_url)
res=requests.get(pdf_url)   

with open(filename, 'wb') as f:    
    f.write(res.content)    
```
