### <span style="color:white"> res=requests.post(url, data=data)  | data = 要傳遞的參數內容 </span>

```python
import pandas as pd
df_fs=pd.read_excel(r"C:\Users\user\Documents\實證研究\python_ds\上市櫃公司.xlsx")
df_fs_code=df_fs["代號"]
```

```python
import requests
from bs4 import BeautifulSoup
import os
save_dir = "年報"
# data = 設定要傳遞的參數內容
data={'id': '',
      'key': '',
      'step': '1',
      'co_id': str(df_fs_code[0]),
      'year': '114',
      'seamon': '',
      'mtype': 'F',
      'dtype': 'F04'}    

url='https://doc.twse.com.tw/server-java/t57sb01'  
# 第一個post請求
res=requests.post(url, data=data) 
soup=BeautifulSoup(res.text, "html.parser") 
link=soup.find('a')    
print("第1個link:",link)
filename=link.text
print("檔案名稱:",filename)

#第二個post請求
data={'step': '9',
      'kind': 'F',
      'co_id': str(df_fs_code[0]),
      'filename': filename} 

res=requests.post(url, data=data)   
soup=BeautifulSoup(res.text, 'html.parser')
link=soup.find('a') 
print("第2個link:",link)
href=link.get('href')
print("超連結:",href) 
pdf_url='https://doc.twse.com.tw' + href   

print("網址:",pdf_url)
res=requests.get(pdf_url)  
# 2024_2330_20250603F04.pdf -> ["2024_2330", "20250603F04.pdf"] 1=分割次數
filename = filename.rsplit('_', 1)[0] + ".pdf"
filename = os.path.join(save_dir, filename)
# python 用來開啟檔案的函式
with open(filename, 'wb') as f:    
    f.write(res.content)    
```
